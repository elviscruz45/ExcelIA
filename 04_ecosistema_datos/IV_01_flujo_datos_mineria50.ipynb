## 2. Concepto

### OT vs IT
- **OT (Operational Technology):** PLC, DCS, sensores, actuadores — tiempo real.
- **IT (Information Technology):** bases de datos, reportes, dashboards — análisis y gestión.

### Modelo Purdue (simplificado)
| Nivel | Ejemplo en concentradora |
|-------|--------------------------|
| 0 | Sensores: temperatura, vibración, presión |
| 1 | PLC / control bomba y molino |
| 2 | SCADA / supervisión de área |
| 3 | PI System historiador, MES |
| 4 | Supabase, data lake, Python ETL |
| 5 | Power BI, ERP mantenimiento |

### Flujo de datos (línea molienda)
```
Sensor → PLC → PI Historian → Python (ETL) → Supabase → Power BI → Alerta mantenimiento
```

## 4. Herramientas en la cadena de valor

| Herramienta | Fortaleza | Limitación |
|-------------|-----------|------------|
| **Excel** | Modelos locales, KPIs rápidos | Escala, automatización, versionado |
| **SQL/Supabase** | Consultas, historial centralizado | Requiere modelado de datos |
| **Python** | ETL, ML, integración | Curva de aprendizaje |
| **Power BI** | Dashboards ejecutivos | Dependiente de calidad de datos |
