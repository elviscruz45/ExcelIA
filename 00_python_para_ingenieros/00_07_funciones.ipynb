# Lección 00_07 — Funciones

## 1. Objetivo

Crear funciones reutilizables para cálculos de mantenimiento: disponibilidad, clasificación de vibración y conversión de unidades.

## 2. Concepto

Una **función** encapsula lógica con nombre, parámetros y valor de retorno. Evita repetir el mismo código en cada análisis de turno.

### Si vienes de Excel...

Una **función personalizada** o **macro** en Excel es similar a `def mi_funcion():` en Python. La ventaja: puedes combinarlas en pipelines de análisis.

## 3. Ejemplos guiados

```python
import os
from pathlib import Path

MOD_DIR = Path.cwd()
os.chdir(MOD_DIR)
OUTPUT_DIR = MOD_DIR / "outputs"
OUTPUT_DIR.mkdir(exist_ok=True)
DATA_DIR = MOD_DIR / "data"
```

```python
def calcular_disponibilidad(mtbf_horas, mttr_horas):
    """Disponibilidad = MTBF / (MTBF + MTTR)"""
    return mtbf_horas / (mtbf_horas + mttr_horas)

disp_pump101 = calcular_disponibilidad(mtbf_horas=500, mttr_horas=12)
print(f"Disponibilidad PUMP101: {disp_pump101:.1%}")
```

```python
def clasificar_vibracion(rms, umbral_alerta=4.5, umbral_critico=7.1):
    if rms >= umbral_critico:
        return "Critico"
    elif rms >= umbral_alerta:
        return "Alerta"
    return "Normal"

# Reutilizar la función para varios equipos
equipos_vib = {"PUMP101": 3.2, "PUMP102": 5.8, "PUMP103": 7.5}
for equipo, rms in equipos_vib.items():
    print(f"{equipo}: {clasificar_vibracion(rms)}")
```

```python
def bar_a_psi(presion_bar):
    """Convierte presión de bar a psi."""
    return presion_bar * 14.5038

presion_succion = 2.5  # bar
print(f"Presión succión: {presion_succion} bar = {bar_a_psi(presion_succion):.1f} psi")
```

## 5. Ejercicio práctico — Función de eficiencia de bomba

Crea `calcular_eficiencia(caudal_medido, caudal_diseno)` que devuelva el porcentaje. Pruébala con caudal medido 115 y diseño 130 m3/h.

```python
def calcular_eficiencia(caudal_medido, caudal_diseno):
    return (caudal_medido / caudal_diseno) * 100

eficiencia = calcular_eficiencia(115, 130)
print(f'Eficiencia: {eficiencia:.1f}%')
```

## 6. Resumen y siguiente paso

- `def nombre(parametros):` define una función.
- `return` devuelve el resultado.
- Las funciones evitan duplicar lógica en análisis repetitivos.
