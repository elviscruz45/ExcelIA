# Lección 00_03 — Operadores y cadenas de texto

## 1. Objetivo

Realizar cálculos de ingeniería (potencia, caudal), comparar valores con umbrales y formatear mensajes de alerta para reportes de turno.

## 2. Concepto

Los **operadores** permiten calcular y comparar. Las **f-strings** (`f"..."`) insertan variables en texto para generar alertas legibles.

### Si vienes de Excel...

`=A1*B1` en Excel equivale a `corriente_a * tension_v` en Python. `=A1>70` devuelve VERDADERO/FALSO; en Python es `temp > 70`.

## 3. Ejemplos guiados

```python
import os
from pathlib import Path

MOD_DIR = Path.cwd()
os.chdir(MOD_DIR)
OUTPUT_DIR = MOD_DIR / "outputs"
OUTPUT_DIR.mkdir(exist_ok=True)
DATA_DIR = MOD_DIR / "data"
```

```python
# Potencia del motor del molino: P = I × V
corriente_a = 320
tension_v = 6600
potencia_kw = (corriente_a * tension_v) / 1000
print(f"Potencia estimada: {potencia_kw:,.0f} kW")
```

```python
# Comparaciones con umbrales de vibración
vibracion_rms = 5.2
umbral_alerta = 4.5
print("¿Supera alerta?", vibracion_rms > umbral_alerta)
print("¿Dentro de rango normal?", vibracion_rms <= umbral_alerta)
```

```python
# f-string para alerta de mantenimiento
equipo = "PUMP101"
temp_rodamiento_c = 78.5
umbral_critico_c = 80

mensaje = f"ALERTA {equipo}: temperatura rodamiento {temp_rodamiento_c} °C (umbral {umbral_critico_c} °C)"
print(mensaje)
```

```python
# Reporte de turno con múltiples variables
turno = "Noche"
operador = "J. Pérez"
paradas_h = 1.5

reporte = (
    f"Reporte turno {turno} | Operador: {operador} | "
    f"Paradas no programadas: {paradas_h} h"
)
print(reporte)
```

## 5. Ejercicio práctico — Alerta de presión de descarga

La bomba PUMP102 tiene presión de descarga 9.8 bar y umbral mínimo 10 bar. Crea variables y un f-string que indique si hay riesgo de cavitación.

```python
equipo = "PUMP102"
presion_bar = 9.8
umbral_min_bar = 10.0
riesgo = presion_bar < umbral_min_bar
print(f"{equipo}: presión {presion_bar} bar | Riesgo cavitación: {riesgo}")
```

## 6. Resumen y siguiente paso

- Operadores `+ - * /` para cálculos; `> < >= <=` para comparar.
- f-strings formatean alertas: `f"Temp {t} °C"`.
- Las comparaciones devuelven `True` o `False`.
